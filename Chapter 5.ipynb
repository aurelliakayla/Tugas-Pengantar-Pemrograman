```python
import pandas as pd #untuk membaca data dari excell
from scipy.stats.mstats import gmean, hmean #untuk menghitung geometric mean dan harmonic mean
import matplotlib.pyplot as plt #untuk membuat histogram dan menampilkan garis vertikal untuk masing-masing mean

my_dataset = pd.read_excel('Smith_glass_post_NYT_data.xlsx', sheet_name = 'Supp_traces') #membaca data dari file excel dan menyimpan dalam variabel my_dataset


a_mean = my_dataset.Rb.mean() #menghitung arithmetic mean dari kolom Rb dalam dataset
g_mean = gmean(my_dataset['Rb']) #menghitung geometric mean dari kolom Rb dalam dataset
h_mean = hmean(my_dataset['Rb']) # menghitung harmonic mean dari kolom Rb dalam dataset

print ('-------') #mencetak garis pemisah
print ('arithmetic mean') #mencetak judul untuk arithmetic mean
print ("{0:.1f} [ppm]".format(a_mean)) #mencetak nilai arithmetic mean dengan format satu angka di belakang komasatu ppm
print ('-------') #mencetak garis pemisah

print ('geometric mean') #mencetak judul untuk geometric mean
print ("{0:.1f} [ppm]".format(g_mean)) #mencetak nilai arithmetic mean dengan format satu angka di belakang komasatu ppm
print ('-------') #mencetak garis pemisah

print ('harmonic mean') #mencetak judul untuk harmonic mean
print ("{0:.1f} [ppm]".format(h_mean)) #mencetak nilai arithmetic mean dengan format satu angka di belakang komasatu ppm
print ('-------') #mencetak garis pemisah

fig, ax = plt.subplots() #membuat figure dan axes untuk plot
ax.hist(my_dataset.Rb, bins='auto', density=True, edgecolor='k', label='Measurements Hist', alpha=0.8) #membuat histogram dari kolom Rb dalam dataset dengan jumlah bin otomatis, menormalkan histogram, memberikan warna tepi pada setiap bin, memberikan label untuk legenda, dan mengatur transparansi histogram
ax.axvline(a_mean, color='purple', label='Arithmetic mean', linewidth=2) #menambahkan garis vertikal pada posisi arithmetic mean dengan warna ungu, label untuk legenda, dan ketebalan garis 2
ax.axvline(g_mean, color='pink',  label='Geometric mean', linewidth=2) #menambahkan garis vertikal pada posisi geometric mean dengan warna merah muda, label untuk legenda, dan ketebalan garis 2
ax.axvline(h_mean, color='orange',  label='Harmonic mean', linewidth=2) #menambahkan garis vertikal pada posisi harmonic mean dengan warna orange, label untuk legenda, dan ketebalan garis 2
ax.set_xlabel('Rb [ppm]') #mengatur label untuk sumbu x
ax.set_ylabel('Probability density')  # mengatur label untuk sumbu y
ax.legend() # menampilkan legenda untuk garis vertikal yang menunjukkan masing-masing mean

'''
Output:
-------
arithmetic mean
365.4 [ppm]
-------
geometric mean
348.6 [ppm]
-------
harmonic mean
333.8 [ppm]
-------
'''
```

```python
import pandas as pd #untuk membaca data dari excell
import matplotlib.pyplot as plt #untuk membuat histogram dan menampilkan garis vertikal untuk masing-masing mean

my_dataset = pd.read_excel('Smith_glass_post_NYT_data.xlsx', sheet_name='Supp_traces') #membaca data dari file excel dan menyimpan dalam variabel my_dataset

variance =  my_dataset['Y'].var() #menghitung variance dari kolom Y dalam dataset
stddev =  my_dataset['Y'].std() # menghitung standard deviation dari kolom Y dalam dataset

print ('-------') #mencetak garis pemisah
print ('Variance') #mencetak judul untuk variance
print("{0:.0f} [square ppm]".format(variance)) #mencetak nilai variance dengan format tanpa angka di belakang koma dan satuan square ppm
print ('-------') #mencetak garis pemisah
print ('Standard Deviation') #mencetak judul untuk standard deviation
print("{0:.0f} [ppm]".format(stddev)) #mencetak nilai standard deviation dengan format tanpa angka di belakang koma dan satuan ppm
print ('-------') #mencetak garis pemisah

fig, ax = plt.subplots() #membuat figure dan axes untuk plot
ax.hist(my_dataset.Y, bins= 20, density = True, edgecolor='k', label='Measurements Hist') #membuat histogram dari kolom Y dalam dataset dengan jumlah bin 20, menormalkan histogram, memberikan warna tepi pada setiap bin, dan memberikan label untuk legenda
ax.axvline(my_dataset['Y'].mean()-stddev, color='pink', label=r'mean - 1$\sigma$', linewidth=2) #menambahkan garis vertikal pada posisi mean - 1 standard deviation dengan warna merah muda, label untuk legenda yang menunjukkan mean - 1 sigma, dan ketebalan garis 2
ax.axvline(my_dataset['Y'].mean()+stddev, color='green', label=r'mean + 1$\sigma$', linewidth=2) #menambahkan garis vertikal pada posisi mean + 1 standard deviation dengan warna hijau, label untuk legenda yang menunjukkan mean + 1 sigma, dan ketebalan garis 2
ax.axvspan(my_dataset['Y'].mean()-stddev, my_dataset['Y'].mean()+stddev, alpha=0.1, color='orange', label=r'mean $\pm$ 1$\sigma$') #menambahkan area vertikal yang diarsir dengan warna oranye antara mean - 1 standard deviation dan mean + 1 standard deviation, dengan transparansi 0.1, dan label untuk legenda yang menunjukkan mean ± 1 sigma
ax.set_xlabel('Y [ppm]') #mengatur label untuk sumbu x
ax.set_ylabel('Probability density') #mengatur label untuk sumbu y
ax.legend() #menampilkan legenda untuk garis vertikal yang menunjukkan mean - 1 sigma, mean + 1 sigma, dan area yang menunjukkan mean ± 1 sigma

'''
Output:
-------
Variance
14021 [square ppm]
-------
Standard Deviation
118 [ppm]
-------
'''
```

```python
import pandas as pd #untuk membaca data dari excell
import numpy as np #untuk menghitung histogram dan bin edges
import matplotlib.pyplot as plt #untuk membuat histogram dan menampilkan garis vertikal untuk masing-masing mean

my_dataset = pd.read_excel('Smith_glass_post_NYT_data.xlsx', sheet_name='Supp_traces') #membaca data dari file excel dan menyimpan dalam variabel my_dataset

q1 = np.percentile(my_dataset.Cs, 25, method='midpoint') #menghitung kuartil pertama (Q1) dari kolom Cs dalam dataset dengan menggunakan method 'midpoint'
q3 = np.percentile(my_dataset.Cs, 75, method= 'midpoint') #menghitung kuartil pertama (Q3) dari kolom Cs dalam dataset dengan menggunakan method 'midpoint'
  
iqr = q3 - q1 # Interquaritle range (IQR)  

print ('-------') #mencetak garis pemisah
print ('Interquaritle range (IQR): {0:.0f} [ppm]'.format(iqr)) #mencetak nilai interquartile range (IQR) dengan format tanpa angka di belakang koma dan satuan ppm
print ('-------') #mencetak garis pemisah

fig, ax = plt.subplots() #membuat figure dan axes untuk plot
ax.hist(my_dataset.Cs, bins='auto', density=True, edgecolor='k', label='Measurements Hist') #membuat histogram dari kolom CS dalam dataset dengan bin otomatis, menormalkan histogram, memberikan warna tepi pada setiap bin, dan memberikan label untuk legenda
ax.axvline(q1, color='purple', label='Q1', linewidth=2) #menambahkan garis vertikal pada posisi Q1 dengan warna ungu, label untuk legenda, dan ketebalan garis 2
ax.axvline(q3, color='khaki', label='Q3', linewidth=2)  # menambahkan garis vertikal pada posisi Q3 dengan warna khaki, label untuk legenda, dan ketebalan garis 2
ax.axvspan(q1, q3, alpha=0.1, color='orange', label='Interquaritle range (IQR)') #menambahkan area vertikal yang diarsir dengan warna oranye antara Q1 dan Q3, dengan transparansi 0.1, dan label untuk legenda
ax.set_xlabel('Cs [ppm]') #mengatur label untuk sumbu x
ax.set_ylabel('Probability density') #mengatur label untuk sumbu y
ax.legend() #menampilkan legenda untuk garis vertikal yang menunjukkan Q1, Q3, dan area yang menunjukkan IQR

'''
Output:
-------
Interquaritle range (IQR): 164 [ppm]
-------
'''
```

```python
import pandas as pd #untuk membaca data dari file excel
import numpy as np #untuk menghitung histogram dan bin edges
import matplotlib.pyplot as plt #untuk membuat histogram dan menampilkan garis vertikal untuk masing masing mean

my_dataset = pd.read_excel('Smith_glass_post_NYT_data.xlsx', sheet_name='Supp_traces') #membaca data dari file excel dan menyimpan dalam variabel my_dataset

a_mean = my_dataset.Ba.mean() #menghitung arithmetic mean dari kolom Ba dalam dataset

median = my_dataset.Ba.median() #menghitung median dari kolom BA dalam dataset

hist, bin_edges = np.histogram(my_dataset['Ba'], bins=20, density=True) #menghitung histogram dan bin edges dari kolom Ba dalam dataset dengan jumlah bin 20 dan menormalkan histogram
modal_value = (bin_edges[hist.argmax()] +  bin_edges[hist.argmax()+1])/2 #menghitung modal value dengan mengambil rata-rata dari bin edges yang memiliki frekuensi tertinggi dalam histogram

fig, ax = plt.subplots() #membuat figure dan axes untuk plot
ax.hist(my_dataset.Ba, bins=20, density=True, edgecolor='k', label="Measurements Hist") #membuat histogram dari kolom Ba dalam dataset dengan jumlah bin 20, menormalkan histogram, memberikan warna tepi pada setiap bin, dan memberikan label untuk legenda
ax.axvline(modal_value, color='brown', label='Modal Value', linewidth=2) #menambahkan garis vertikal pada posisi modal value dengan warna coklat, label untuk legenda, dan ketebalan garis 2
ax.axvline(median, color='black',  label='Median Value', linewidth=2) #menambahkan garis vertikal pada posisi median value dengan warna hitam, label untuk legenda, dan ketebalan garis 2 
ax.axvline(a_mean, color='grey',  label='Arithmetic mean', linewidth=2) #menambahkan garis vertikal pada posisi arithmetic mean dengan warna abu-abu, label untuk legenda, dan ketebalan garis 2 
ax.set_xlabel('Ba [ppm]') #mengatur label untuk sumbu x
ax.set_ylabel('Probability density') #mengatur label untuk sumbu y
ax.legend() #menampilkan legenda untuk garis vertikal yang menunjukkan modal value, median value, dan arithmetic mean

```

```python
import numpy as np #untuk menghitung histogram dan bin edges
from scipy.stats import skew #untuk menghitung skewness dari kolom Pr dalam dataset

a_mean = my_dataset.Pr.mean() #menghitung arithmetic mean dari kolom Pr dalam dataset
median = my_dataset.Pr.median() #menghitung median dari kolom Pr dalam dataset
hist, bin_edges = np.histogram(my_dataset['Pr'], bins=20, density=True) #menghitung histogram dan bin edges dari kolom Pr dalam dataset dengan jumlah bin 20 dan menormalkan histogram
modal_value = (bin_edges[hist.argmax()] + bin_edges[hist.argmax()+1])/2 #menghitung modal value dengan mengambil rata-rata dari bin edges yang memiliki frekuensi tertinggi dalam histogram
standard_deviation = my_dataset['Pr'].std() #menghitung standard deviation dari kolom Pr dalam dataset

a1 = (a_mean - modal_value) / standard_deviation #menghitung Pearson's first coefficient of skewness
a2 = 3 * (a_mean - median) / standard_deviation #menghitung Pearson's 2nd moment of skewness
g1 = skew(my_dataset['Pr']) #menghitung Fisher-Pearson's coefficient of skewness dengan menggunakan fungsi skew dari scipy.stats yang menghitung skewness dari kolom Pr dalam dataset

print ('-------') #mencetak garis pemisah
print ("Pearson's first coefficient of skewness: {:.2f}".format(a1)) #mencetak nilai Pearson's first coefficient of skewness dengan format dua angka di belakang koma
print ("Pearson's 2nd moment of skewness: {:.2f}".format(a2)) #mencetak nilai Pearson's 2nd moment of skewness dengan format dua angka di belakang koma
print ("Fisher-Pearson's coefficient of skewness: {:.2f}".format(g1)) #mencetak nilai Fisher-Pearson's coefficient of skewness dengan format dua angka di belakang koma
print ('-------') #mencetak garis pemisah

'''
Output:
-------
Pearson's first coefficient of skewness: 0.74
Pearson's 2nd moment of skewness: 0.66
Fisher-Pearson's coefficient of skewness: 1.26
-------
'''
```

```python
import pandas as pd  #untuk membaca data dari file excel
import matplotlib.pyplot as plt #untuk membuat histogram dan menampilkan garis vertikal untuk masing masing mean

my_dataset = pd.read_excel('Smith_glass_post_NYT_data.xlsx', sheet_name='Supp_traces') #membaca data dari file excel dan menyimpan dalam variabel my_dataset

fig, ax = plt.subplots() # membuat figure dan axes untuk plot
my_flierprops = dict(markerfacecolor='#f8e9a1', markeredgecolor='#24305e', marker='o') #mendefinisikan properti untuk outliers pada boxplot dengan menggunakan marker lingkaran berwarna kuning muda dengan tepi berwarna biru tua
my_medianprops = dict(color='#f76c6c', linewidth = 2) #mendefinisikan properti untuk garis median pada boxplot dengan menggunakan warna merah muda dan ketebalan garis 2
my_boxprops = dict(facecolor='#a8d0e6', edgecolor='#24305e') #mendefinisikan properti untuk box pada boxplot dengan menggunakan warna biru muda untuk isi box dan warna biru tua untuk tepi box
ax.boxplot(my_dataset.Sm, patch_artist = True, notch=True, flierprops=my_flierprops, medianprops=my_medianprops, boxprops=my_boxprops) #membuat boxplot untuk kolom Sm dalam dataset dengan menggunakan properti yang telah didefinisikan untuk outliers, median, dan box, serta mengaktifkan opsi notch untuk menampilkan lekukan pada boxplot
ax.set_ylabel('Sm [ppm]') #mengatur label untuk sumbu y
ax.set_xticks([1]) #mengatur posisi tick pada sumbu x untuk menampilkan satu tick pada posisi 1
ax.set_xticklabels(['all Epochs'])  #mengatur label untuk tick pada sumbu x dengan teks 'all Epochs' untuk menunjukkan bahwa boxplot mencakup semua data dari berbagai epoch
plt.show() #menampilkan plot boxplot yang telah dibuat
```
