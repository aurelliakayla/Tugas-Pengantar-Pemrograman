```python
import pandas as pd  # untuk data handling
import matplotlib.pyplot as plt  # membuat figure dan plotting

my_dataset = pd.read_excel(
    'Smith_glass_post_NYT_data.xlsx',
    sheet_name='Supp_traces'
)  # membaca data dari file Excel

# filter data berdasarkan Epoch
my_dataset1 = my_dataset[my_dataset['Epoch'] == 'one'] # Memfilter dataset untuk epoch 'one' dan menyimpannya dalam variabel my_dataset1
my_dataset2 = my_dataset[my_dataset['Epoch'] == 'two']

fig = plt.figure()  # membuat figure baru

ax1 = fig.add_subplot(2, 1, 1) # Menambahkan subplot pada posisi 1 dari grid 2x1

ax1.scatter(
    my_dataset1['Zr'], my_dataset1['Dy'],
    marker='s', color="#f4e9c7",
    edgecolor='#000000',
    label="First Epoch"
) # Plot scatter untuk kolom 'Zr' dan 'Th' dengan marker 's', warna hex RGB '#c7ddf4', edgecolor '#000000', dan label "First Epoch"

ax1.scatter(
    my_dataset2['Zr'], my_dataset2['Dy'],
    marker='o', color="#46fff0",
    edgecolor='#000000',
    label="Second Epoch"
)

ax1.set_xlabel("Zr [ppm]") # Mengatur label sumbu x
ax1.set_ylabel("Dy [ppm]")
ax1.legend(
    loc='upper left',
    framealpha=1,
    frameon=True,
    title="Age < 15 ky",
    title_fontsize=10
) # Menampilkan legenda dengan lokasi 'upper left', latar belakang tidak transparan, frame diaktifkan, dan judul legenda "Age < 15 ky" dengan ukuran font 10

ax2 = fig.add_subplot(2, 1, 2)

ax2.scatter(
    my_dataset1['Zr'], my_dataset1['Dy'],
    marker='s', color="#c9f4c7",
    edgecolor='#000000',
    label="First Epoch"
)

ax2.scatter(
    my_dataset2['Zr'], my_dataset2['Dy'],
    marker='o', color="#f69fd3",
    edgecolor='#000000',
    label="Second Epoch"
)

ax2.set_xlabel("Zr [ppm]")
ax2.set_ylabel("Dy [ppm]")
ax2.legend(
    frameon=False,
    loc='lower right',
    ncol=2,
    title="Age < 15 ky",
    title_fontsize=10
)

fig.tight_layout() # Menyesuaikan tata letak subplot agar tidak saling tumpang tindih
plt.show() # Menampilkan plot
```

```python
# Introductory examples
name = 'Alya' # Mendefinisikan variabel name dengan string 'Alya'
surname = 'Pratama'

print('-------------------------------------------------') # Mencetak garis pemisah
print('My name is {}'.format(name)) # Mencetak string dengan format yang menyisipkan nilai variabel name menggunakan metode format()
print('-------------------------------------------------')
print('My name is {} and my surname is {}'.format(name, surname))
print('-------------------------------------------------')

# Decimal Number formatting
PI = 3.14159265358979323846 # Mendefinisikan variabel PI dengan nilai Archimedes' constant

print('----------------------------------------------------')
print("The 2 digit Archimedes' constant is equal to {:.2f}".format(PI)) # Mencetak string dengan format yang menyisipkan nilai variabel PI dengan 2 digit desimal menggunakan metode format()
print("The 3 digit Archimedes' constant is equal to {:.3f}".format(PI))
print("The 4 digit Archimedes' constant is equal to {:.4f}".format(PI))
print("The 5 digit Archimedes' constant is equal to {:.5f}".format(PI))
print('----------------------------------------------------')

'''Results 
-------------------------------------------------
My name is Alya
-------------------------------------------------
My name is Alya and my surname is Pratama
-------------------------------------------------
----------------------------------------------------
The 2 digit Archimedes' constant is equal to 3.14
The 3 digit Archimedes' constant is equal to 3.142
The 4 digit Archimedes' constant is equal to 3.1416
The 5 digit Archimedes' constant is equal to 3.14159
----------------------------------------------------
'''
```

```python
# Explicit positive and negative reporting
a = +7.84231 # Mendefinisikan variabel a dengan nilai positif 7.84231
b = -3.918276 

print('-------------------------------------------------') # Mencetak string dengan format yang menyisipkan nilai variabel a dan b dengan 2 digit desimal, tanpa menampilkan simbol plus untuk nilai positif menggunakan metode format()
print("The plus symbol is not reported: {:.2f} | {:.2f}".format(+7.84231, -3.918276)) # Mencetak string dengan format yang menyisipkan nilai variabel a dan b dengan 2 digit desimal, tanpa menampilkan simbol plus untuk nilai positif menggunakan metode format()
print("The plus symbol is reported: {:+.2f} | {:+.2f}".format(a, b))
print('-------------------------------------------------')

# Reporting as percent
c = 0.2845

print('-------------------------------------------------')
print("Reporting as percent: {:.1%}".format(c))
print('-------------------------------------------------')

# Scientific notation
d = 9820000000000

print('-------------------------------------------------')
print("Scientific notation using e: {:.1e}".format(d))
print("Scientific notation using E: {:.1E}".format(d))
print('-------------------------------------------------')

'''Results
-------------------------------------------------
The plus symbol is not reported: 7.84 | -3.92
The plus symbol is reported: +7.84 | -3.92
-------------------------------------------------
-------------------------------------------------
Reporting as percent: 28.5%
-------------------------------------------------
-------------------------------------------------
Scientific notation using e: 9.8e+12
Scientific notation using E: 9.8E+12
-------------------------------------------------
'''
```

```python
import pandas as pd # untuk data handling
import matplotlib.pyplot as plt # Membuat figure dan axes untuk plotting
import numpy as np # untuk operasi matematika


def my_line(x, m, q): # Mendefinisikan fungsi my_line yang mengambil tiga argumen: x, m, dan q
    return m * x + q # Fungsi mengembalikan hasil dari m dikalikan dengan x ditambah q


my_dataset = pd.read_excel(
    'Smith_glass_post_NYT_data.xlsx',
    sheet_name='Supp_majors',
    engine='openpyxl'
) # membaca data dari file Excel dengan menggunakan engine 'openpyxl' untuk memastikan kompatibilitas dengan format file Excel yang lebih baru

my_dataset1 = my_dataset[my_dataset['Epoch'] == 'one'] # Memfilter dataset untuk epoch 'one' dan menyimpannya dalam variabel my_dataset1
my_dataset2 = my_dataset[my_dataset['Epoch'] == 'two']

x = np.linspace(52.5, 62, 100) # Membuat array x yang terdiri dari 100 nilai yang terdistribusi secara merata antara 52.5 dan 62 menggunakan fungsi linspace() dari NumPy
y = my_line(x, m=0.3, q=-10.3)

fig, ax = plt.subplots()

ax.scatter(
    my_dataset1['SiO2'], my_dataset1['K2O'],
    marker='s', color="#c7f4cc",
    edgecolor='#000000',
    label=r'$1^{st}$ Epoch'
) # Plot scatter untuk kolom 'SiO2' dan 'K2O' dengan marker 's', warna hex RGB '#c7ddf4', edgecolor '#000000', dan label "1st Epoch"

ax.scatter(
    my_dataset2['SiO2'], my_dataset2['K2O'],
    marker='s', color="#ff4681",
    edgecolor='#000000',
    label=r'$2^{nd}$ Epoch'
)

ax.plot(x, y, color="#2a5777")

ax.annotate(
    r'What is the 1$\sigma$ for this point?',
    xy=(47.6, 6.6),
    xytext=(47, 8.8),
    arrowprops=dict(arrowstyle="->", connectionstyle="arc3")
) # Menambahkan anotasi pada plot dengan teks "What is the 1$\sigma$ for this point?", posisi titik yang dianotasi (47.6, 6.6), posisi teks (47, 8.8), dan properti panah yang menunjukkan arah dari titik ke teks

ax.text(
    52.4, 5.6,
    r'$Na_2O = 0.3 \cdot SiO_2 -10.3$',
    fontsize=10,
    rotation=33
) # Menambahkan teks pada plot dengan posisi (52.4, 5.6), isi teks berupa persamaan linear, ukuran font 10, dan rotasi 33 derajat

ax.text(
    53.5, 5.1,
    r'$\mu_{SiO_2} = \frac{a_1+a_2+\cdots+a_n}{n} = $'
    + '{:.1f} [wt.%]'.format(57.721),
    fontsize=11.5
)

ax.set_xlabel(r'SiO$_2$ [wt%]') # Mengatur label sumbu x dengan format LaTeX untuk menampilkan subscript pada SiO2
ax.set_ylabel(r'K$_2$O [wt%]')

ax.legend() # Menampilkan legenda dengan pengaturan default

plt.show() # Menampilkan plot
```

```python
import matplotlib.pyplot as plt
import numpy as np


def ec(f, d, c0):
    cl = c0 / (d * (1-f) + f)
    return cl


my_f = np.linspace(0.1, 1, 10)

my_c1 = ec(f=my_f, d=0.1, c0=100)

colors = ['#ff9494', '#cbeaa2', '#d1a396', '#828fc3', '#95b2e5', '#e9b8f4', '#f4b8e5', '#b8f4f2', '#c5f4b8', '#f9ca78']

fig = plt.figure()
ax1 = fig.add_subplot(2, 2, 1)
ax1.plot(my_f, my_c1, marker='o', linestyle='-', markersize=5)
ax1.set_xlabel('F')
ax1.set_ylabel('C [ppm]')

ax2 = fig.add_subplot(2, 2, 2)
ax2.scatter(my_f, my_c1, marker='o', s=my_f*150)
ax2.set_xlabel('F')
ax2.set_ylabel('C [ppm]')

ax3 = fig.add_subplot(2, 2, 3)
ax3.scatter(my_f, my_c1, marker='o', c=colors, s=my_f*150)
ax3.set_xlabel('F')
ax3.set_ylabel('C [ppm]')

ax4 = fig.add_subplot(2, 2, 4)
ax4.plot(my_f, my_c1, marker='', linestyle='-', zorder=0)
ax4.scatter(my_f, my_c1, marker='o', c=colors, s=my_f*150, zorder=1)
ax4.set_xlabel('F')
ax4.set_ylabel('C [ppm]')

fig.tight_layout()
```

```python
import matplotlib.pyplot as plt # Membuat figure dan axes untuk plotting
import numpy as np # untuk operasi matematika


def ec(f, d, c0): # Mendefinisikan fungsi ec yang mengambil tiga argumen: f, d, dan c0
    return c0 / (d * (1 - f) + f) # Fungsi mengembalikan hasil dari c0 dibagi dengan (d dikalikan dengan (1 - f) ditambah f)


my_f = np.linspace(0.1, 1, 10) # Membuat array my_f yang terdiri dari 10 nilai yang terdistribusi secara merata antara 0.1 dan 1 menggunakan fungsi linspace() dari NumPy

my_c1 = ec(f=my_f, d=0.1, c0=100) # Menghitung nilai my_c1 dengan memanggil fungsi ec() dan menyisipkan array my_f sebagai argumen f, 0.1 sebagai argumen d, dan 100 sebagai argumen c0

colors = [
    '#ff9494', '#cbeaa2', '#d1a396', '#828fc3',
    '#95b2e5', '#e9b8f4', '#f4b8e5', '#b8f4f2',
    '#c5f4b8', '#f9ca78'
] # Mendefinisikan daftar warna hex RGB untuk digunakan dalam plot

fig = plt.figure() # Membuat figure baru

ax1 = fig.add_subplot(2, 2, 1) # Menambahkan subplot pada posisi 1 dari grid 2x2
ax1.plot(my_f, my_c1, marker='o', linestyle='-', markersize=5) # Plot garis dengan marker 'o', linestyle '-', dan ukuran marker 5 untuk array my_f dan my_c1 pada subplot ax1
ax1.set_xlabel('F') # Mengatur label sumbu x
ax1.set_ylabel('C [ppm]')

ax2 = fig.add_subplot(2, 2, 2)
ax2.scatter(my_f, my_c1, marker='o', s=my_f * 150)
ax2.set_xlabel('F')
ax2.set_ylabel('C [ppm]')

ax3 = fig.add_subplot(2, 2, 3)
ax3.scatter(my_f, my_c1, marker='o', c=colors, s=my_f * 150)
ax3.set_xlabel('F')
ax3.set_ylabel('C [ppm]')

ax4 = fig.add_subplot(2, 2, 4)
ax4.plot(my_f, my_c1, linestyle='-', zorder=0)
ax4.scatter(my_f, my_c1, marker='o', c=colors, s=my_f * 150, zorder=1)
ax4.set_xlabel('F')
ax4.set_ylabel('C [ppm]')

fig.tight_layout() # Menyesuaikan tata letak subplot agar tidak saling tumpang tindih
plt.show() # Menampilkan plot
```
