```python
print('a sequence from 0 to 2') # Mencetak judul untuk urutan angka dari 0 hingga 2
for i in range(3): # Perulangan untuk setiap angka dalam rentang 0 hingga 2 (3 tidak termasuk)
    print(i) # Mencetak angka saat ini dalam perulangan
    
print('----------------------')  # Mencetak garis pemisah 
print('a sequence from 2 to 4')
for i in range(2, 5):
    print(i)

print('----------------------')    
print('a sequence from 2 to 8 with a step of 2')
for i in range(2, 9, 2):
    print(i)
    
'''
Output:
a sequence from 0 to 2 # Karena rentang dimulai dari 0 dan berakhir sebelum 3, maka instruksi ini akan dieksekusi untuk angka 0, 1, dan 2
0
1
2
----------------------
a sequence from 2 to 4
2
3
4
----------------------
a sequence from 2 to 8 with a step of 2
2
4
6
8
'''
```

```python
def sum(a, b):# Mendefinisikan fungsi dengan nama sum yang menerima dua parameter a dan b
    c = a + b # Menjumlahkan nilai a dan b, lalu menyimpan hasilnya dalam variabel c
    return c # Mengembalikan nilai c sebagai hasil dari fungsi sum

res = sum(a=2, b=3) # Memanggil fungsi sum dengan argumen a=2 dan b=3, lalu menyimpan hasilnya dalam variabel res
print('the result is ' + str(res)) # Mencetak hasil penjumlahan dengan mengkonversi nilai res menjadi string untuk digabungkan dengan teks

'''
Output:
the result is 5 # Karena fungsi sum menjumlahkan 2 dan 3, maka hasilnya adalah 5, yang kemudian dicetak sebagai bagian dari string.
'''
```
